```python
import pandas as pd
import os
import json

# กำหนด Path
data_path = os.path.join('project', 'data', 'samples', 'financial_mockup_data.csv')
schema_path = os.path.join('project', 'data', 'schemas', 'financial_schema.json')

# โหลดข้อมูล
if os.path.exists(data_path):
    df = pd.read_csv(data_path, parse_dates=['timestamp'])
    print(f"📊 Loaded {len(df)} records.")
else:
    print("❌ ไม่พบไฟล์ข้อมูล! กรุณารัน src/mockup_data.py ก่อน")
```

```python
# โหลด Schema มาเช็ค Column
with open(schema_path, 'r') as f:
    schema = json.load(f)

expected_columns = [field['name'] for field in schema['fields']]
missing_cols = set(expected_columns) - set(df.columns)

if not missing_cols:
    print("✅ Data structure is valid.")
else:
    print(f"⚠️ Missing columns: {missing_cols}")

# เช็คค่าว่าง (Missing Values)
print("\nMissing Values Check:")
print(df.isnull().sum())
```
